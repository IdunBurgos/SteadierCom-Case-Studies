# Run subprocess

```python
import subprocess
import time
import select
```

```python
with open("err_file.txt",'w') as errfile:
    process = subprocess.Popen("../scripts_bash/steadiercom_parallel_gut.sh", 
                               shell=True, 
                               text=True, 
                               executable="/bin/bash", 
                               stdout=subprocess.PIPE, 
                               stderr=errfile,
                               )

    while True:
        line = process.stdout.readline()
        if not line:
            break
        print(line.rstrip(), flush=True)
```
